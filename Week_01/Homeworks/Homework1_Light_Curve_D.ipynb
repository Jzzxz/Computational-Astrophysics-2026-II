# Homework 01 — Light Curve: Taylor Series Convergence

## Problem

A light curve can be modeled as

$$
y(t)=y_0+A\sin(\phi+\omega t).
$$

For simplicity, taking $y_0=0$ and the phase angle $\phi=0$ gives

$$
y(t)=A\sin(\omega t).
$$

The sine function can be represented using its Taylor series as

$$
A\sin(\omega t)
=
A\omega t
\sum_{k=0}^{\infty}
\frac{(-1)^k(\omega t)^{2k}}{(2k+1)!}.
$$

For this problem, consider

$$
A=0.1,
\qquad
\omega=3.105,
\qquad
t=1,
\qquad
\Delta t=0.2.
$$

## Task

Implement the Taylor series expansion in Python and determine the value of $k$ required to obtain **7 correct significant figures**.

The convergence of successive approximations will be evaluated using the approximate relative error

$$
\varepsilon_a
=
\left|
\frac{S_k-S_{k-1}}{S_k}
\right|100,
$$

where $S_k$ represents the Taylor series truncated at $k$.

The required accuracy will be determined using the **Scarborough criterion**:

$$
\varepsilon_s
=
0.5\times10^{2-m}\%,
$$

where $m$ is the desired number of significant figures. For this problem,

$$
m=7.
$$

```python
import math
#Parametros
A = 0.1
w = 3.105
t=1
delta_t= 0.2
```

```python
#Punto d
error_scarborough_punto_d = 0.5 * 10**(2-7)
k_d = 1
error_aprox_d = 1000

while error_aprox_d > error_scarborough_punto_d:
  suma_pasada = sum(((-1)**k * (w*t)**(2*k)) / math.factorial(1+ 2*k) for k in range(k_d))
  suma_actualizada = sum(((-1)**k * (w*t)**(2*k)) / math.factorial(1+ 2*k) for k in range(k_d + 1))
  error_aprox_d = abs((suma_actualizada-suma_pasada)/suma_actualizada) *100
  k_d += 1

print("k =", k_d-1)
print("Error aproximado=", error_aprox_d)
print("Criterio de Scarborough para 7 cifras=", error_scarborough_punto_d)
```

```python
#Confirmemos que para k=9 no se tenia:

suma_pasada_9 = sum(((-1)**k * (w*t)**(2*k)) / math.factorial(1+ 2*k) for k in range(9))
suma_actualizada_9 = sum(((-1)**k * (w*t)**(2*k)) / math.factorial(1+ 2*k) for k in range(10))
error_k9 = abs((suma_actualizada_9 - suma_pasada_9) / suma_actualizada_9) * 100

print("Error para k=9 =", error_k9)
print("Criterio =", error_scarborough_punto_d)
```
