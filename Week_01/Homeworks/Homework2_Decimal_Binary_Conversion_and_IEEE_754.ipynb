## Tasks

**a)** Apply the implemented functions to obtain the binary representations of

$$
18.0308, \qquad 26.0985, \qquad 94.0516.
$$

Use the reference conversion above to verify the implementation.

````python
def entero_a_binario(n):
  if n == 0:
    return "0"
  binario = ""
  while n > 0:
    residuo = n % 2
    binario = str(residuo) + binario
    n = n // 2
  return binario

def fraccion_a_binario(n, max_bits = 20):
  binario_decimal = ""
  binario_entero = entero_a_binario(int(n//1))
  parte_decimal = n - (n//1)
  contador = 0
  while parte_decimal != 0 and contador < max_bits:
    parte_decimal = parte_decimal * 2
    if parte_decimal > 1:
      binario_decimal += "1"
      parte_decimal = parte_decimal - 1
    elif parte_decimal==1:
      binario_decimal += "1"
      break
    else:
      binario_decimal += "0"
    contador += 1
  return str(binario_entero)+"."+str(binario_decimal)
````

**d)** — IEEE 754 Float32 Encoding

Using the functions developed in class in the previous parts of the exercise, implement a function in **pure Python** that follows the **IEEE 754 encoding algorithm for 32-bit floating-point numbers (`float32`)**.

Apply the function to the same decimal numbers considered in part **(a)**:

$$
18.0308, \qquad 26.0985, \qquad 94.0516.
$$

For each number, obtain its IEEE 754 `float32` representation and compare the results with the reference functions presented in class.

````python
def ieee754_float32(n):

    # Bit de signo
    if n < 0:
        signo = "1"
        n = -n
    else:
        signo = "0"

    numero_binario = fraccion_a_binario(n, max_bits=50)
    binario_entero, binario_decimal = numero_binario.split(".")
    exponente = len(binario_entero) - 1
    mantisa_completa = binario_entero[1:] + binario_decimal

    while len(mantisa_completa) < 24:
        mantisa_completa += "0"

    mantisa = mantisa_completa[:23]
    bit_redondeo = mantisa_completa[23]

    if bit_redondeo == "1":

        mantisa_entero = 0

        for bit in mantisa:
            mantisa_entero = mantisa_entero * 2 + int(bit)

        mantisa_entero += 1

        if mantisa_entero == 2**23:
            mantisa = "0" * 23
            exponente += 1
        else:
            mantisa = entero_a_binario(mantisa_entero)

            while len(mantisa) < 23:
                mantisa = "0" + mantisa

    exponente_bias = exponente + 127
    exponente_binario = entero_a_binario(exponente_bias)

    while len(exponente_binario) < 8:
        exponente_binario = "0" + exponente_binario

    resultado = signo + exponente_binario + mantisa
    return resultado
````

````python
# Punto d
numeros = [18.0308, 26.0985, 94.0516]

for numero in numeros:

    resultado = ieee754_float32(numero)

    print("Número:", numero)
    print("Signo:", resultado[0])
    print("Exponente:", resultado[1:9])
    print("Mantisa:", resultado[9:])
    print("IEEE 754 float32:", resultado)
    print()
````
