# Exercise 02 — Light Curve and Taylor Series

The phase curve of a light curve can be modeled as

$$
y(t) = y_0 + A\sin(\phi + \omega t).
$$

For simplicity, let $y_0 = 0$ and the phase angle $\phi = 0$. Therefore,

$$
y(t) = A\sin(\omega t).
$$

Using the Taylor series expansion of the sine function, the light curve can be approximated by

$$
A\sin(\omega t)
=
A\omega t
\sum_{k=0}^{\infty}
\frac{(-1)^k(\omega t)^{2k}}{(2k+1)!}.
$$

Consider the following parameters:

$$
A = 0.1, \qquad
\omega = 3.105, \qquad
t = 1, \qquad
\Delta t = 0.2.
$$
