# Exercise 03 — Taylor Series Truncation Error

## Problem

Consider the Taylor series expansion of the exponential function

$$
e^x
=
\sum_{k=0}^{\infty}\frac{x^k}{k!}.
$$

When the series is truncated after $n+1$ terms, the truncation error is given by the remainder

$$
R_n(x)
=
\frac{f^{(n+1)}(\xi)x^{n+1}}{(n+1)!},
\qquad
0\leq \xi \leq x.
$$

For the exponential function, $f^{(n+1)}(\xi)=e^\xi$.

Using the first **4 terms** of the Taylor series of $e^x$ and taking

$$
x=0.5,
$$

complete the following tasks:

**a)** Write a Python function to approximate $e^x$ using the first 4 terms of its Taylor series.

```python
import math
#Punto a
#Recordemos la Taylor de e es como sum de x^k/k! escribamosla, pero para los primeros 4 termino
x=0.5
aprox_e = sum((x**k)/math.factorial(k) for k in range(4))
aprox_e
```

**b)** Compare the approximation with `math.exp(0.5)` and calculate the **true error** and the **true relative error**.


```python
#Punto b
exacto_e = math.exp(x)
print(exacto_e)
error_verdadero = abs(exacto_e-aprox_e)
print(error_verdadero)
error_relativo = abs(exacto_e-aprox_e)/exacto_e*100
print(f"{error_relativo} %")
```

**c)** Use the remainder formula $R_n(x)$ to obtain a theoretical upper bound for the truncation error. Recall that

$$
0<\xi<0.5
$$

and that

$$
e^\xi < e^{0.5} < 2,
$$

which provides a useful upper bound.

**Punto c)**

Estamos truncando los primeros 4 terminos osea $n+1 = 4$, asi queda el residuo, como $0<\epsilon<0.5$ y elevando esta
desigualdad por exp, teniendo en cuenta que la funcion es creciente y mantiene el orden de las desigualdades entonces
$1 < e^{\epsilon} < e^{0.5} < 2$ entonces $|R_3|< \frac{2 (0.5)^4}{4!}$, calculemoslo:

```python
#Punto c

R_3 = (2 * 0.5**4) / math.factorial(4)
R_3
```
