# Homework 02 — Composite Simpson's 1/3 Rule n = 4

## Problem

The vertical distance traveled by a rocket from $t=8$ to $t=30$ seconds is given by

$$
x=
\int_{8}^{30}
\left(
2000\ln\left[\frac{140000}{140000-2100t}\right]
-9.8t
\right)dt.
$$

1. Use the **composite Simpson's 1/3 rule** with $n=4$ segments to estimate the distance traveled between $t=8$ and $t=30$ seconds.
2. Compute the **absolute error** using the exact value of the integral.
3. Compute the **relative error** using the exact value of the integral.

## Theoretical Idea

The composite Simpson's 1/3 rule divides the interval $[a,b]$ into an even number $n$ of equally spaced segments.

The step size is

$$
h=\frac{b-a}{n},
$$

and the nodes are

$$
x_i=a+ih,
\qquad i=0,1,\ldots,n.
$$

The composite Simpson's 1/3 approximation is

$$
\int_a^b f(x)\,dx
\approx
\frac{h}{3}
\left[
f(x_0)
+
4\sum_{\substack{i=1\\i\text{ odd}}}^{n-1}f(x_i)
+
2\sum_{\substack{i=2\\i\text{ even}}}^{n-2}f(x_i)
+
f(x_n)
\right].
$$

Since

$$
h=\frac{b-a}{n},
$$

the formula can also be written as

$$
\int_a^b f(x)\,dx
\approx
\frac{b-a}{3n}
\left[
f(x_0)
+
4\sum_{\substack{i=1\\i\text{ odd}}}^{n-1}f(x_i)
+
2\sum_{\substack{i=2\\i\text{ even}}}^{n-2}f(x_i)
+
f(x_n)
\right].
$$

The interior points with odd indices receive a weight of $4$, while the interior points with even indices receive a weight of $2$.

The absolute and relative errors are

$$
E_{\mathrm{abs}}
=
|I_{\mathrm{exact}}-I_{\mathrm{Simpson}}|,
$$

and

$$
E_{\mathrm{rel}}
=
\frac{|I_{\mathrm{exact}}-I_{\mathrm{Simpson}}|}
{|I_{\mathrm{exact}}|}.
$$

```python
import numpy as np

def f(t):
  return 2000*np.log(140000/(140000-2100*t))-9.8*t
```

```python
#Punto a)

#Regla de Simpson compuesta n=4
t_0 = 8
t_final = 30
n = 4
h = (t_final - t_0)/n
integral = 11/6*(f(t_0)+ 4*(f(t_0+1*h)+f(t_0+3*h))+ 2*f(t_0+2*h)+f(t_0+4*h))

print(f"La integral aproximada por regla de Simpson compuesta con n=4 es: {integral}")
```

```python
#Punto b) y c)
integral_exacta = 11061.335

error_absoluto = abs(integral - integral_exacta)
error_relativo = error_absoluto/integral_exacta*100

print(f"El error absoluto es: {error_absoluto}")
print(f"El error relativo es: {error_relativo}%")
```
