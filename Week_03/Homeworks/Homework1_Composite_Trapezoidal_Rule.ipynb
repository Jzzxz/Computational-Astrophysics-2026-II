# Homework 01 — Composite Trapezoidal Rule

## Problem

The vertical distance traveled by a rocket from $t=8$ to $t=30$ seconds is given by

$$
x=
\int_{8}^{30}
\left(
2000\ln\left[\frac{140000}{140000-2100t}\right]
-9.8t
\right)dt.
$$

1. Use the **composite trapezoidal rule** with $n=2$ segments to estimate the distance traveled.
2. Compute the **absolute error** using the exact value of the integral.
3. Compute the **relative error** using the exact value of the integral.
4. Construct a table containing the numerical approximation, absolute error, and relative error for

$$
n=1,2,3,4,5,6,7,8.
$$

## Theoretical Idea

The composite trapezoidal rule divides the interval $[a,b]$ into $n$ segments of equal width

$$
h=\frac{b-a}{n}.
$$

The integral is then approximated by applying the trapezoidal rule to each subinterval:

$$
\int_a^b f(x)\,dx
\approx
\frac{h}{2}
\left[
f(a)
+
2\sum_{i=1}^{n-1}f(a+ih)
+
f(b)
\right].
$$

As the number of segments $n$ increases, the step size $h$ decreases and the numerical approximation is expected to approach the exact value of the integral.

The absolute and relative errors are

$$
E_{\mathrm{abs}}=|I_{\mathrm{exact}}-I_T|,
$$

and

$$
E_{\mathrm{rel}}
=
\frac{|I_{\mathrm{exact}}-I_T|}{|I_{\mathrm{exact}}|}.
$$

```python
import numpy as np

def funcion_cohete(t):
  return 2000*np.log(140000/(140000-2100*t))-9.8*t

def trapecio_compuesto(f,a,b,n):

  h = (b-a)/n

  suma = 0

  for i in range(1,n):
    suma += f(a+i*h)

  integral = (h/2)*(f(a)+2*suma+f(b))

  return integral

#Punto a)
a = 8
b = 30
n = 2

x_trapecio = trapecio_compuesto(funcion_cohete,a,b,n)
print("Distancia aproximada =",x_trapecio)

#Punto b)
x_exacto = 11061.33
print("Distancia exacta =",x_exacto)
error_absoluto = abs(x_exacto-x_trapecio)
print("Error absoluto =",error_absoluto)

#Punto c)
error_relativo = error_absoluto/abs(x_exacto)*100
print("Error relativo =",error_relativo,"%")

#Punto d)
n_valores = []
x_valores = []
errores_absolutos = []
errores_relativos = []

for n in range(1,9):
  x = trapecio_compuesto(funcion_cohete,a,b,n)
  error_abs = abs(x_exacto-x)
  error_rel = error_abs/abs(x_exacto)
  n_valores.append(n)
  x_valores.append(x)
  errores_absolutos.append(error_abs)
  errores_relativos.append(error_rel)

print("n     x               Error absoluto      Error relativo")
for i in range(len(n_valores)):
  print(n_valores[i],x_valores[i],errores_absolutos[i],errores_relativos[i])
```
