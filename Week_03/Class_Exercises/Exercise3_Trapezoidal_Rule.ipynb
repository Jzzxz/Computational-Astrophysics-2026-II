# Exercise 03 — Trapezoidal Rule

## Problem

The vertical distance traveled by a rocket from $t=8$ to $t=30$ seconds is given by

$$
x=
\int_{8}^{30}
\left(
2000\ln\left[\frac{140000}{140000-2100t}\right]
-9.8t
\right)dt.
$$

1. Use the **trapezoidal rule** to estimate the distance traveled between $t=8$ and $t=30$ seconds.
2. Compute the **absolute error** of the approximation.
3. Compute the **relative error** of the approximation.

## Theoretical Idea

The trapezoidal rule approximates a definite integral by replacing the function between two points with a straight line. The area under the curve is then approximated by the area of a trapezoid.

For an integral

$$
\int_a^b f(x)\,dx,
$$

the trapezoidal approximation is

$$
I_T=\frac{b-a}{2}\left[f(a)+f(b)\right].
$$

The absolute and relative errors are

$$
E_{\mathrm{abs}}=|I_{\mathrm{exact}}-I_T|,
$$

and

$$
E_{\mathrm{rel}}
=
\frac{|I_{\mathrm{exact}}-I_T|}{|I_{\mathrm{exact}}|}.
$$

```python
import numpy as np
#Regla del trapecio para aproximar integrales

def funcion(t):
  return 2000*np.log(140000/(140000-2100*t)) - 9.81*t
a = 8
b = 30
M = 1000
x = np.linspace(a,b,M)
h = (b-a)/(M-1)
suma = 0
for i in range(M):
  suma += funcion(x[i])
integral = h*((funcion(x[0])+funcion(x[-1]))/2 + np.sum(funcion(x[1:-1])))

#Solo punto inicial y final (TRAPECIO)

integral_1trapecio = (b-a)*(funcion(x[-1])+funcion(x[0]))/2

#Integral analitica

integral_analitica = 11057.16

#Resultados
print(f"La integral analitica es: {integral_analitica}")
print(f"La integral aproximada por regla del trapecio y 2 nodos es: {integral_1trapecio}")
print(f"La integral aproximada por regla del trapecio y 1000 nodos es: {integral}")

#Error absoluto
print(f"El error absoluto por regla del trapecio y 2 nodos es: {abs(integral_1trapecio-integral_analitica)}")
print(f"El error absoluto por regla del trapecio y 1000 nodos es: {abs(integral-integral_analitica)}")

#Error relativo
print(f"El error relativo por regla del trapecio y 2 nodos es: {(abs(integral_1trapecio-integral_analitica))/integral_analitica}")
print(f"El error relativo por regla del trapecio y 1000 nodos es: {(abs(integral-integral_analitica))/integral_analitica}")
```
