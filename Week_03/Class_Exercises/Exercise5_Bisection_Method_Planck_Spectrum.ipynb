# Exercise 05 — Bisection Method: Planck Spectrum

## Problem

The wavelength $\lambda_{\max}$ at which a blackbody emits the maximum amount of radiation can be obtained from Planck's radiation law.

After differentiating the spectral distribution and introducing the dimensionless variable

$$
x=\frac{hc}{\lambda k_B T},
$$

the problem reduces to finding the non-zero root of

$$
f(x)=x-5(1-e^{-x})=0.
$$

Use the **Bisection Method** to determine the root of this equation.

Use the initial interval

$$
x_l=0.5,
\qquad
x_u=5,
$$

and the Scarborough criterion for at least three significant figures.

The stopping criterion is therefore

$$
\varepsilon_s=0.5\times10^{2-3}\%=0.05\%.
$$

Report the numerical approximation of the root and evaluate $f(x)$ at the obtained value.

## Theory

The Bisection Method is a root-finding method for continuous functions. It starts with an interval $[x_l,x_u]$ such that the function changes sign:

$$
f(x_l)f(x_u)<0.
$$

If this condition is satisfied, at least one root exists inside the interval.

At each iteration, the midpoint of the current interval is calculated as

$$
x_r=\frac{x_l+x_u}{2}.
$$

The sign of the function is then used to determine which half of the interval contains the root.

If

$$
f(x_l)f(x_r)<0,
$$

the root is located in the left half of the interval, so

$$
x_u=x_r.
$$

Otherwise, the root is located in the right half and

$$
x_l=x_r.
$$

This process is repeated, reducing the size of the interval by half at every iteration.

To estimate the numerical error, the relative approximate error between two consecutive approximations is calculated as

$$
\varepsilon_a=
\left|
\frac{x_r^{(i)}-x_r^{(i-1)}}
{x_r^{(i)}}
\right|100\%.
$$

The Scarborough criterion specifies the stopping error required to guarantee at least $n$ significant figures:

$$
\varepsilon_s=0.5\times10^{2-n}\%.
$$

For three significant figures,

$$
\varepsilon_s=0.05\%.
$$

Therefore, the iterations continue while

$$
\varepsilon_a>0.05\%.
$$

For the Planck spectrum problem, the function to solve is

$$
f(x)=x-5(1-e^{-x}).
$$

The Bisection Method progressively reduces the initial interval $[0.5,5]$ until the approximate relative error satisfies the required tolerance.

```python
import numpy as np

def espectro_planck(x):
  return x-5*(1-np.exp(-x))
```

```python
#Punto a) y b)

def biseccion(f, x_izq, x_der):
  error = 1000
  while error > 5e-2:
    if f(x_izq)*f(x_der)<0:
      x_new = (x_izq+x_der)/2
      if f(x_izq)*f(x_new)<0:
        error = (abs(x_der - x_new))/abs(x_new)*100
        x_der = x_new
      else:
        error = (abs(x_izq - x_new))/abs(x_new)*100
        x_izq = x_new
  return x_new
```

```python
def f(x):
  return x**3-20
```

```python
biseccion(f,1,4)
```

```python
#Punto c)
print(f"La raiz esta en: {biseccion(espectro_planck,0.5,5)} y da: {espectro_planck(biseccion(espectro_planck,0.5,5))}")
```

```python
def biseccion2(f,x_izq, x_der):
  error = 1000
  x_anterior = None

  while error>5e-2:
    x_new = (x_izq+x_der)/2
    if x_anterior is not None:
      error = abs((x_new-x_anterior)/x_new)*100
    if f(x_izq)*f(x_new) < 0:
      x_der = x_new
    else:
      x_izq = x_new
    x_anterior = x_new
  return x_new
```

```python
raiz = biseccion2(espectro_planck, 0.5, 5)
```

```python
print(f"La raiz es {raiz} y da {espectro_planck(raiz)} ")
```
