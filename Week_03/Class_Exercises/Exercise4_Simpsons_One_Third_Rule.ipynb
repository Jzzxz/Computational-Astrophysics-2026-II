# Exercise 04 — Simpson's 1/3 Rule

## Problem

Consider the integral

$$
\int_{0.1}^{1.3} 5x e^{-2x}\,dx.
$$

1. Use **Simpson's 1/3 rule** to estimate the value of the integral.
2. Compute the **absolute error** using the exact value of the integral.
3. Compute the **relative error** using the exact value of the integral.

## Theoretical Idea

Simpson's 1/3 rule approximates the integrand using a second-degree polynomial passing through three equally spaced points.

For an interval $[a,b]$, the midpoint is

$$
x_1=\frac{a+b}{2},
$$

and the step size is

$$
h=\frac{b-a}{2}.
$$

Simpson's 1/3 rule is given by

$$
\int_a^b f(x)\,dx
\approx
\frac{h}{3}
\left[
f(a)+4f(x_1)+f(b)
\right].
$$

Equivalently,

$$
\int_a^b f(x)\,dx
\approx
\frac{b-a}{6}
\left[
f(a)+4f\left(\frac{a+b}{2}\right)+f(b)
\right].
$$

The absolute and relative errors are

$$
E_{\mathrm{abs}}
=
|I_{\mathrm{exact}}-I_{\mathrm{Simpson}}|,
$$

and

$$
E_{\mathrm{rel}}
=
\frac{|I_{\mathrm{exact}}-I_{\mathrm{Simpson}}|}
{|I_{\mathrm{exact}}|}.
$$

```python
import numpy as np
#Regla de Simpson

def funcion2(x):
  return 5*x*np.exp(-2*x)

a1= 0.1
b1 = 1.3
h = (b1-a1)/2
integral_simpson_13 = h/3*(funcion2(a1) + 4*funcion2((a1+b1)/2) + funcion2(b1))
integral_analitica2 = 0.894865

print(f"La integral analitica es: {integral_analitica2}")
print(f"La integral por simpson es: {integral_simpson_13}")
print(f"El error absoluto de la integral por simpson es: {abs(integral_simpson_13-integral_analitica2)}")
print(f"El error relativo de la integral por simpson es: {abs(integral_simpson_13-integral_analitica2)/integral_analitica2*100}%")
```
