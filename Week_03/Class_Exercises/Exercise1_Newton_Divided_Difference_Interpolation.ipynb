# Exercise 01 — Newton Divided Difference Interpolation

## Problem

The upward velocity of a rocket is given as a function of time by the following data:

| $t$ (s) | $v(t)$ (m/s) |
|---:|---:|
| 0 | 0 |
| 10 | 227.04 |
| 15 | 362.78 |
| 20 | 517.35 |
| 22.5 | 602.97 |
| 30 | 901.67 |

Using **second-order Newton divided difference interpolation**, determine the velocity of the rocket at

$$
t=21\ \text{s}.
$$

Implement the Newton interpolation method in Python and use the appropriate data points to estimate

$$
v(21).
$$

## Theoretical Background

Newton's divided difference interpolation constructs a polynomial from a set of known data points. For a second-order interpolation using three points, the polynomial is

$$
P_2(x)=a_0+a_1(x-x_0)+a_2(x-x_0)(x-x_1),
$$

where the coefficients are obtained from the divided differences:

$$
a_0=f(x_0),
$$

$$
a_1=\frac{f(x_1)-f(x_0)}{x_1-x_0},
$$

$$
a_2=
\frac{
\frac{f(x_2)-f(x_1)}{x_2-x_1}
-
\frac{f(x_1)-f(x_0)}{x_1-x_0}
}{
x_2-x_0
}.
$$

## Solution Approach

To estimate the rocket velocity at $t=21$ s, the three nearby data points at $t=15$, $20$, and $22.5$ s are selected. These points are used to calculate the Newton divided differences and construct the quadratic interpolation polynomial. Finally, the polynomial is evaluated at $t=21$ s.

```python
import numpy as np

t = [0, 10, 15, 20, 22.5, 30]
v = [0, 227.04, 362.78, 517.35, 602.97, 901.67]

#Queremos aproximacion cuadratica en t=21
#Tomamos t[2], t[3], t[4] que son los mas cercanos

#Queremos encontrar p(21) = a0+a1(x-t[2])+a2(x-t[2])(x-t[3])

a0 = v[2]
a1 = (v[3]-v[2])/(t[3]-t[2])
auxiliar = (v[4]-v[3])/(t[4]-t[3])
a2 = (auxiliar-a1)/(t[4]-t[2])

x = 21

p = a0 + a1*(x-t[2]) + a2*(x-t[2])*(x-t[3])
print(f"La aproximacion cuadratica en t = 21 es: {p}")
```
