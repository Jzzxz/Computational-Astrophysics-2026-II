# Exercise 01 — Supernova Luminosity: Finite Differences

## Problem

Consider the following model for the luminosity decay of a supernova:

$$
L(t)=L_0e^{-t/\tau},
$$

where $L_0$ is the initial luminosity and $\tau$ is the characteristic decay time.

For this exercise, consider

$$
L_0 = 5.0\times10^{42},
\qquad
\tau = 111.4\ \text{days},
$$

and evaluate the numerical derivatives at

$$
t_0 = 64\ \text{days},
\qquad
h=\Delta t=2\ \text{days}.
$$

The finite difference approximations for the first derivative are

$$
D_{\mathrm{forward}}
=
\frac{L(t_0+h)-L(t_0)}{h},
$$

$$
D_{\mathrm{backward}}
=
\frac{L(t_0)-L(t_0-h)}{h},
$$

and

$$
D_{\mathrm{centered}}
=
\frac{L(t_0+h)-L(t_0-h)}{2h}.
$$

Implement in Python the functions

```python
L(t)
diff_forward()
diff_backward()
diff_centered()
diff2_centered()
```

## Tasks

**a)** Calculate the exact derivative $L'(t)$ and evaluate it at the specified time.

````python
import numpy as np

#Parametros
L_0 = 5.0*10**(42)
tau = 111.4
t_0 = 64
h = 2

#Funcion
def L(t):
 return L_0*np.exp(-t/tau)

#Derivada analitica
def derivada_analitica(t):
    return -(L_0/tau)*np.exp(-t/tau)

#Punto a
derivada_exacta_0 = derivada_analitica(0)
derivada_exacta_64 = derivada_analitica(64)

print(f"La derivada exacta en t = 0 es: {derivada_exacta_0}")
print(f"La derivada exacta en t = 64 es: {derivada_exacta_64}")
````

**b)** Calculate the first derivative using the **forward**, **backward**, and **centered finite difference** approximations.

````python
def derivada_forward(t):
  return (L(t+h)- L(t))/h
print(f"La derivada con diferencias finitas hacia adelante es: {derivada_forward(64)}")
def derivada_backward(t):
  return (L(t) - L(t-h))/h
print(f"La derivada con diferencias finitas hacia atras es: {derivada_backward(64)}")
def derivada_centrada(t):
  return (L(t+h) -  L(t-h))/(2*h)
print(f"La derivada con diferencias finitas centradas es: {derivada_centrada(64)}")
````

**c)** Calculate the percentage relative error of each finite difference approximation with respect to the exact derivative.

````python
print(f"Error relativo hacia adelante: {abs((derivada_exacta_64-derivada_forward(64))/derivada_exacta_64) *100}%")
print(f"Error relativo hacia atras: {abs((derivada_exacta_64-derivada_backward(64))/derivada_exacta_64) * 100}%")
print(f"Error relativo centrada: {abs((derivada_exacta_64-derivada_centrada(64))/derivada_exacta_64) * 100}%")
````

Vemos hacia adelante y atras converge linealmente mientras que centrada
converge cuadraticamente, como indica la teoria.
