# Homework 01 — Second Derivative: Finite Differences

## Problem

Using the Taylor series expansion of $f(x)$, derive the finite difference approximations for the **second derivative**

$$
f''(x)
$$

using the following schemes:

- Forward finite difference
- Backward finite difference
- Centered finite difference

Then, consider the supernova luminosity decay model

$$
L(t)=L_0e^{-t/\tau},
$$

with

$$
L_0 = 5.0\times10^{42},
\qquad
\tau = 111.4\ \text{days},
$$

and evaluate the second derivative at

$$
t_0 = 64\ \text{days},
\qquad
h=\Delta t=2\ \text{days}.
$$

## Tasks

**a)** Using Taylor series expansions, derive the **forward**, **backward**, and **centered finite difference** formulas for $f''(x)$.

```python
import numpy as np

#Parametros
L_0 = 5.0*10**(42)
tau = 111.4
t_0 = 64
h = 2

#Funcion
def L(t):
  return L_0*np.exp(-t/tau)

#Punto a
def segunda_derivada_forward(f,x,h):
  return (f(x+2*h)-2*f(x+h)+f(x))/(h**2)

def segunda_derivada_backward(f,x,h):
  return (f(x)-2*f(x-h)+f(x-2*h))/(h**2)

def segunda_derivada_centrada(f,x,h):
  return (f(x+h)-2*f(x)+f(x-h))/(h**2)

print(f"Segunda derivada hacia adelante: {segunda_derivada_forward(L,64,2)}")
print(f"Segunda derivada hacia atras: {segunda_derivada_backward(L,64,h)}")
print(f"Segunda derivada centrada: {segunda_derivada_centrada(L,64,h)}")
```

**b)** Calculate the exact second derivative $L''(t)$ and evaluate it at $t_0=64$ days.

```python
#Punto b
def segunda_derivada_analitica(t):
  return (L_0/tau**2)*np.exp(-t/tau)

segunda_derivada_exacta_64 = segunda_derivada_analitica(t_0)

print(f"La segunda derivada exacta en t = 64 es: {segunda_derivada_exacta_64}")
```

**c)** Approximate $L''(t_0)$ using the **centered finite difference** formula.

```python
#Punto c
segunda_derivada_centrada_64 = segunda_derivada_centrada(L,t_0,h)

print(f"La segunda derivada con diferencias finitas centradas en t = 64 es: {segunda_derivada_centrada_64}")
```

**d)** Calculate the percentage relative error of the centered finite difference approximation with respect to the exact value of $L''(t_0)$.

```python
#Punto d

error_relativo_centrada = abs((segunda_derivada_exacta_64-segunda_derivada_centrada_64)/segunda_derivada_exacta_64)*100

print(f"Error relativo de la segunda derivada centrada: {error_relativo_centrada}%")
```

Vemos que sigue siendo orden del error $O(h^2)$ como indica la teoria

## Interpretation of the Derivatives

The **first derivative** measures the instantaneous rate of change of a function. Geometrically, it represents the slope of the tangent line to the function at a given point.

The **second derivative** measures how the first derivative changes. Geometrically, it describes the concavity of the function:

- $f''(x)>0$: concave upward.
- $f''(x)<0$: concave downward.

In the supernova model, $L'(t)$ describes how the luminosity changes with time, while $L''(t)$ describes how the rate of luminosity decay itself changes with time.
